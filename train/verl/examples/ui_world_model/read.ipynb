```python
import pandas as pd

# 读取 Parquet 文件为 DataFrame
df = pd.read_parquet(
    "/path/to/cuwm/project/rl-0.6.0/verl/examples/ui_world_model/no_last/test.parquet",
    engine="pyarrow",
)

source_set = set(source_paths)

# 保留 extra_info['pair_path'] 在 source_paths 里的行
mask = df["extra_info"].apply(lambda x: isinstance(x, dict) and x.get("pair_path") in source_set)
df_keep = df[mask].reset_index(drop=True)

print("before:", len(df), "after:", len(df_keep))
df_keep
```
